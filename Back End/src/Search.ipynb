```python
from sentence_transformers import SentenceTransformer, util
import torch

model = SentenceTransformer("BAAI/bge-m3")  #all-MiniLM-L6-v2, paraphrase-multilingual-MiniLM-L12-v2, BAAI/bge-m3

places = [
    "Quán cà phê view đẹp",
    "Nhà hàng hải sản tươi sống",
    "Quán ăn giá rẻ cho sinh viên",
    "Khách sạn gần biển",
    "Khu du lịch sinh thái"
]

place_embeddings = model.encode(
    places,
    convert_to_tensor=True,
    normalize_embeddings=True
)

query = input("Bạn muốn tìm gì: ")

query_embedding = model.encode(
    query,
    convert_to_tensor=True,
    normalize_embeddings=True
)

scores = util.cos_sim(query_embedding, place_embeddings)[0]

top_k = 3
top_results = torch.topk(scores, k=top_k)

print("\nKết quả phù hợp:\n")

for score, idx in zip(top_results.values, top_results.indices):
    print(f"{places[idx]}  |  Score: {score.item():.4f}")
```
